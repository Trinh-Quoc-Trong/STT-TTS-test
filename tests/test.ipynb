```python

def find_kth_digit(p: int) -> int:
    """
    Finds the p-th digit in the sequence of concatenated positive integers (1234567891011...).

    Args:
        p: The position of the digit to find (1-indexed).

    Returns:
        The digit at the p-th position, or -1 if p is not a positive integer.
    """
    if p <= 0:
        return -1

    num_digits = 1
    count = 9
    # 1. Find how many digits the number containing the p-th digit has.
    while p > num_digits * count:
        p -= num_digits * count
        num_digits += 1
        count *= 10

    # p is now the position within the block of `num_digits`-digit numbers.
    
    # 2. Find the actual number containing the digit.
    # The first number with `num_digits` is 10**(num_digits - 1).
    # The index of our target number within its block is (p - 1) // num_digits.
    index = (p - 1) // num_digits
    start_num = 10**(num_digits - 1)
    number = start_num + index

    # 3. Find the specific digit within that number.
    # The index of the digit within the number is (p - 1) % num_digits.
    digit_index = (p - 1) % num_digits
    
    return int(str(number)[digit_index])

# --- Example Usage ---
# Find the 20,000th digit, as in your original problem.
p = 17
result = find_kth_digit(p)

print(f"The sequence of concatenated integers is: 123456789101112...")
print(f"The digit at position {p} is: {result}")

# Another example: find the 3rd digit
p__2 = 3
result_2 = find_kth_digit(p__2)
print(f"The digit at position {p__2} is: {result_2}")
```
